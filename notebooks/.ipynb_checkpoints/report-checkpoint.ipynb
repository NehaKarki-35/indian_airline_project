## **Flight Delay and Operational Performance Insights Report**
##### Airport Traffic and Flight Scheduling Patterns
- LAX records the highest number of departing flights, followed by DFW and JFK, making them the busiest origin airports in the dataset.
- Flight scheduling is heavily concentrated between 7 AM and 10 AM, with an additional peak observed between 3 PM and 6 PM.
- Very few flights are scheduled during midnight to 5 AM, indicating minimal overnight operations.
- The highest concentration of scheduled departures occurs around 8 AM.
- Overall flight operations are primarily daytime-oriented, with most departures occurring in the morning (6–9 AM) and late afternoon (4–7 PM). Night departures remain relatively rare.
- --

## **Delay Causes and Operational Bottlenecks**
- Carrier Delay and NAS Delay are the primary contributors to total flight delays.
- Weather Delay has a comparatively lower impact on overall delays.
- Security Delay contributes the least to total delays among all delay categories.
- CLT airport records the highest total delays across the dataset, indicating significant operational congestion.
- ---

## Key Delay Patterns and Insights

### Departure vs Arrival Delays
- Departure delay has a strong positive impact on arrival delay.  
- Flights delayed at departure usually arrive late as well.  
- Delay tends to propagate from departure to arrival.  
- Late incoming aircraft creates delays in next flights.  
- Small delays can sometimes be recovered, but large delays are difficult to fix.  
- Airport congestion increases arrival delays.  
- Day of the week has little effect on delays.  

---

### Flight Timing and Delay Distribution
- Evening flights have the highest arrival delays.  
- Morning flights show the lowest delays and better punctuality.  
- Afternoon and evening flights show more variation and outliers.  
- Night flights generally have the least delays.  

---

### Aircraft and System-Level Impact
- Late Aircraft Delay is the main cause of total delays.  
- A delayed incoming aircraft often delays the next departure.  
- Aircraft availability strongly affects delay performance.  
- Delay spreads across flights due to operational dependency.
