### **Literature Review Summary**

Research papers aur airline operations studies me sabse important factors ye mile hain:

- Weather conditions → Delay aur cancellation ka major reason.
- Late arriving aircraft → Chain reaction delays create karta hai.
- Carrier operational issues → Crew, maintenance, scheduling problems.
- Airport congestion (NAS Delay) → Air traffic aur runway congestion.
- Flight distance aur departure time → Delay probability ko affect karte hain.
- Airline/Carrier type → Different airlines ki performance alag hoti hai.
